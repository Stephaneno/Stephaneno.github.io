```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0,10,100)
y=x**2
fig, ax = plt.subplots()
ax.plot (x,y)
ax.plot (2*x,y-2)
ax.set_xlabel("This is X")
ax.set_ylabel("This is Y")
plt.show()
```
