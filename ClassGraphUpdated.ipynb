```python
import pandas as pd
```

```python
chrom_data=pd.read_excel('/content/chrosome_basepairs.xlsx')
```

```python
chrom_data
```

```python
total_basepairs = chrom_data['basepais(bp)'].sum()
print(f"Total basepairs: {total_basepairs}")
```

```python

bp_length = 3.4

chrom_data["Length (Angstrom)"] = chrom_data["basepais(bp)"] * bp_length

chrom_data.head()
```

```python
total_length_angstrom = chrom_data["Length (Angstrom)"].sum()
total_length_meter = total_length_angstrom / 1e10
print(f"Total length of all chromosomes in meters: {total_length_meter} m")
```

```python
total_mbps = total_basepairs / 1_000_000

cost_data = {
    'Year': [2001, 2009, 2010, 2020],
    'Cost per MBps ($)': [10000, 1.00, 0.01, 0.001]
}
cost_df = pd.DataFrame(cost_data)

cost_df['Total Sequencing Cost ($)'] = cost_df['Cost per MBps ($)'] * total_mbps

display(cost_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.barplot(x='Year', y='Total Sequencing Cost ($)', data=cost_df, palette='viridis')
plt.title('Total Whole Genome Sequencing Cost by Year')
plt.xlabel('Year')
plt.ylabel('Total Sequencing Cost ($)')
plt.yscale('log') # Use a log scale due to the large difference in costs
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
